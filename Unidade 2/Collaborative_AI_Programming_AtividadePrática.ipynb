```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    try:
        # Tenta realizar a divisão matemática
        return a / b
    except ZeroDivisionError:
        # Captura especificamente a divisão por zero
        return "Erro: Divisão por zero não é permitida."

def iniciar_calculadora():
    print("--- Calculadora Simples ---")
    try:
        a = float(input("Digite o primeiro número: "))
        b = float(input("Digite o segundo número: "))

        print(f"\nSoma: {somar(a, b)}")
        print(f"Subtração: {subtrair(a, b)}")
        print(f"Multiplicação: {multiplicar(a, b)}")
        print(f"Divisão: {dividir(a, b)}")

    except ValueError:
        # Tratamento para entradas não numéricas (ex: letras)
        print("\nErro: Entrada inválida. Por favor, digite apenas números.")

if __name__ == "__main__":
    iniciar_calculadora()
```

O código atual funciona, mas encerra após um único cálculo. Refatore a função iniciar_calculadora adicionando um loop while para que o usuário possa realizar várias operações em sequência. Crie uma condição de saída clara digitando 'sair'. Mantenha os tratamentos de erro

```python
def somar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Erro: Divisão por zero não é permitida."

def iniciar_calculadora():
    print("--- Calculadora Simples Continua ---")
    print("Dica: Digite 'sair' a qualquer momento no primeiro número para encerrar.")

    while True:
        entrada = input("\nDigite o primeiro número (ou 'sair'): ")

        # Condição de saída avaliada pelo Tech Lead
        if entrada.lower() == 'sair':
            print("Encerrando a calculadora. Até logo!")
            break

        try:
            a = float(entrada)
            b = float(input("Digite o segundo número: "))

            print(f"\nResultados:")
            print(f"Soma: {somar(a, b)}")
            print(f"Subtração: {subtrair(a, b)}")
            print(f"Multiplicação: {multiplicar(a, b)}")
            print(f"Divisão: {dividir(a, b)}")

        except ValueError:
            print("\nErro: Entrada inválida. Digite apenas números ou 'sair'.")

if __name__ == "__main__":
    iniciar_calculadora()
```
